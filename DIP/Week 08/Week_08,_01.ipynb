# **Edge Detection in Image Processing**
**Simple Gradient, Sobel, Laplacian, and Canny**

This notebook introduces basic edge detection techniques used in image processing.

# **TASK 1: LOAD IMAGE**

```python
import urllib.request
import cv2
import matplotlib.pyplot as plt

# Image URL
url = "https://cdn.britannica.com/05/236505-050-17B6E34A/Elon-Musk-2022.jpg"

# Download image
urllib.request.urlretrieve(url, "Elon_musk.jpg")

# Read image in grayscale
image = cv2.imread("Elon_musk.jpg", cv2.IMREAD_GRAYSCALE)

# Show image
plt.imshow(image, cmap='gray')
plt.title("Elon Musk Image")
plt.axis('off')
plt.show()
```

**Explanation:** In this task, we loaded the image using OpenCV and converted it to grayscale. Grayscale is important for edge detection because it simplifies computation.

# **TASK 2: SIMPLE GRADIENT**

```python
gx = np.zeros_like(image)
gy = np.zeros_like(image)

gx[:, :-1] = np.diff(image, axis=1)
gy[:-1, :] = np.diff(image, axis=0)

gradient = np.sqrt(gx**2 + gy**2)

# Display result
plt.imshow(gradient, cmap='gray')
plt.title("Simple Gradient Edge Detection")
plt.axis('off')
plt.show()
```

**Explanation:** This method calculates the difference between neighboring pixels.
 It is the simplest form of edge detection.
 However, it is very sensitive to noise and not very accurate.

# **TASK 3 — Sobel Edge Detection (with Gaussian Blur)**

```python
# Apply Gaussian Blur to reduce noise
blur = cv2.GaussianBlur(image, (5,5), 0)

# Apply Sobel operator
sobel_x = cv2.Sobel(blur, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(blur, cv2.CV_64F, 0, 1, ksize=3)

sobel = np.sqrt(sobel_x**2 + sobel_y**2)

# Display result
plt.imshow(sobel, cmap='gray')
plt.title("Sobel Edge Detection")
plt.axis('off')
plt.show()
```

**Explanation:**
 Sobel operator detects edges using gradients in x and y directions.
Gaussian Blur reduces noise before applying the operator.
This produces more accurate and smoother edges than the simple gradient.

# **TASK 4 — Laplacian Edge Detection**

```python
laplacian = cv2.Laplacian(blur, cv2.CV_64F)

# Display result
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian Edge Detection")
plt.axis('off')
plt.show()

# Explanation:
# Laplacian uses second-order derivatives to detect edges.
# It detects edges in all directions.
# However, it is very sensitive to noise even after smoothing.
```

# **TASK 5 — Canny Edge Detection (Different Thresholds)**

```python
canny1 = cv2.Canny(image, 50, 150)
canny2 = cv2.Canny(image, 100, 200)
canny3 = cv2.Canny(image, 150, 250)

# Display results
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(canny1, cmap='gray')
plt.title("Canny (50,150)")

plt.subplot(1,3,2)
plt.imshow(canny2, cmap='gray')
plt.title("Canny (100,200)")

plt.subplot(1,3,3)
plt.imshow(canny3, cmap='gray')
plt.title("Canny (150,250)")

plt.show()

# Explanation:
# Canny edge detection is a multi-stage algorithm.
# It includes noise reduction, gradient calculation, and thresholding.
# Lower thresholds detect more edges (including noise).
# Higher thresholds detect fewer but cleaner edges.
```

# **TASK 6 — Comparison of All Methods**

```python
plt.figure(figsize=(15,10))

plt.subplot(2,3,1)
plt.imshow(image, cmap='gray')
plt.title("Original")

plt.subplot(2,3,2)
plt.imshow(gradient, cmap='gray')
plt.title("Gradient")

plt.subplot(2,3,3)
plt.imshow(sobel, cmap='gray')
plt.title("Sobel")

plt.subplot(2,3,4)
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian")

plt.subplot(2,3,5)
plt.imshow(canny1, cmap='gray')
plt.title("Canny (50,150)")

plt.subplot(2,3,6)
plt.imshow(canny2, cmap='gray')
plt.title("Canny (100,200)")

plt.tight_layout()
plt.show()

# Explanation:
# This task compares all edge detection methods.
# We can visually analyze which method performs better.
```

# ***FINAL REPORT***

**EDGE DETECTION REPORT**

In this experiment, we implemented and analyzed four different edge detection techniques:
Simple Gradient, Sobel, Laplacian, and Canny.

First, the image "Elon_musk.webp" was loaded and converted to grayscale.
Grayscale simplifies the processing and is essential for edge detection.

The Simple Gradient method was applied using pixel differences.
This method is simple but produces noisy and weak edges.

The Sobel operator was then used with Gaussian Blur preprocessing.
This improved the results significantly by reducing noise and producing clearer edges.

The Laplacian method detected edges in all directions using second-order derivatives.
However, it remained sensitive to noise.

Finally, the Canny edge detection algorithm was applied with different threshold values.
Canny produced the best results due to its multi-stage approach, including noise reduction,
gradient detection, non-maximum suppression, and double thresholding.

**Conclusion:**
Among all methods, Canny edge detection provided the most accurate and clean results.
Simple Gradient was the least effective, while Sobel offered a good balance between simplicity and performance.
